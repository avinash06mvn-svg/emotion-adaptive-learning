```python
#importing libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display plots in notebook
%matplotlib inline

# Better-looking plots
plt.style.use('ggplot')

print("Libraries imported successfully!")
```

```python
#loading and inspecting dataset
df = pd.read_csv("xAPI-Edu-Data.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
#Preview the data
df.head()
```

```python
#Check dataset information
df.info()
```

```python
#Check missing values
df.isnull().sum()
```

```python
#Statistical summary
df.describe()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x='Class')

plt.title("Distribution of Student Performance")
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x='gender')

plt.title("Gender Distribution")
plt.show()
```

```python
plt.figure(figsize=(12,5))

sns.countplot(data=df, x='Topic')

plt.xticks(rotation=45)

plt.title("Students Across Topics")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df, x='StageID')

plt.title("Stage Distribution")

plt.show()
```

```python
engagement = ['raisedhands',
              'VisITedResources',
              'AnnouncementsView',
              'Discussion']

df[engagement].hist(figsize=(10,8))

plt.tight_layout()

plt.show()
```

```python
numeric_df = df.select_dtypes(include='number')

plt.figure(figsize=(6,5))

sns.heatmap(numeric_df.corr(),
            annot=True,
            cmap='Blues')

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.boxplot(data=df,
            x='Class',
            y='raisedhands')

plt.title("Raised Hands vs Student Performance")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.boxplot(data=df,
            x='Class',
            y='VisITedResources')

plt.title("Visited Resources vs Student Performance")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.boxplot(data=df,
            x='Class',
            y='Discussion')

plt.title("Discussion Participation vs Performance")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df,
              x='ParentAnsweringSurvey',
              hue='Class')

plt.title("Parent Survey vs Student Performance")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df,
              x='StudentAbsenceDays',
              hue='Class')

plt.title("Absence vs Student Performance")

plt.show()
```

```python
#very important
engagement = df[['raisedhands',
                 'VisITedResources',
                 'AnnouncementsView',
                 'Discussion']]

engagement.plot(kind='box',
                figsize=(8,5))

plt.title("Student Engagement Features")

plt.show()
```
